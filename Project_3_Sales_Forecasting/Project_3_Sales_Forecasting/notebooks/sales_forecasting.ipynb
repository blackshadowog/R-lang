# Project #3 — Sales Forecasting
Time-series analysis and 12-month sales forecasting using Holt-Winters.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import mean_absolute_error, mean_squared_error
from statsmodels.tsa.holtwinters import ExponentialSmoothing
```

```python
df=pd.read_csv('../data/monthly_sales.csv',parse_dates=['Date'])
df.head()
```

```python
sales=df.set_index('Date')['Sales'].asfreq('MS')
train=sales.iloc[:-12]; test=sales.iloc[-12:]
model=ExponentialSmoothing(train,trend='add',seasonal='mul',seasonal_periods=12,initialization_method='estimated').fit()
forecast=model.forecast(12)
print('MAE:',mean_absolute_error(test,forecast))
print('RMSE:',np.sqrt(mean_squared_error(test,forecast)))
```

```python
plt.figure(figsize=(11,5)); plt.plot(train,label='Train'); plt.plot(test,label='Actual'); plt.plot(forecast,label='Forecast'); plt.legend(); plt.title('Actual vs Forecast Sales'); plt.show()
```
