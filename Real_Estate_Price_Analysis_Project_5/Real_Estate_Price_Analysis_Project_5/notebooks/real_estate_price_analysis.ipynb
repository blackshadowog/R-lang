# 🏠 Real Estate Price Analysis — Project #5
Intermediate analysis of property prices, localities, BHK, area, and price per square foot.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/real_estate.csv')
df.head()
```

```python
print('Shape:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values:')
print(df.isnull().sum())
print('\nDuplicates:', df.duplicated().sum())
```

```python
df.describe(include='all')
```

```python
top5 = df.nlargest(5, 'Price')[['Property_ID','Locality','BHK','Area_sqft','Price']]
bottom5 = df.nsmallest(5, 'Price')[['Property_ID','Locality','BHK','Area_sqft','Price']]
print('Top 5 properties')
display(top5)
print('Bottom 5 properties')
display(bottom5)
```

```python
locality_price = df.groupby('Locality')['Price'].mean().sort_values(ascending=False)
locality_price.plot(kind='bar', figsize=(12,5), title='Average Property Price by Locality')
plt.ylabel('Average Price'); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

```python
bhk_price = df.groupby('BHK')['Price'].mean()
bhk_price.plot(kind='bar', figsize=(8,5), title='Average Price by BHK')
plt.xlabel('BHK'); plt.ylabel('Average Price'); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df, x='Area_sqft', y='Price', hue='BHK')
plt.title('Area vs Property Price')
plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(df[['BHK','Area_sqft','Rate_per_sqft','Price','Property_Age']].corr(), annot=True, fmt='.2f')
plt.title('Correlation Matrix'); plt.tight_layout(); plt.show()
```

```python
rera_price = df.groupby('RERA')['Price'].mean()
print(rera_price)
rera_price.plot(kind='bar', title='Average Price by RERA Status')
plt.ylabel('Average Price'); plt.tight_layout(); plt.show()
```

```python
print('Most expensive locality by average price:', locality_price.index[0])
print('Highest average rate/sqft locality:', df.groupby('Locality')['Rate_per_sqft'].mean().idxmax())
print('Average property price:', round(df['Price'].mean(), 2))
print('Average area:', round(df['Area_sqft'].mean(), 2))
```
